# Сравниваем тексты

На прошлом занятии мы рассмотрели, как преобразовывать текст в вектор. Теперь мы можем сравнивать тексты, сравнивая их векторы. Но как нам это делать?

### Как сравнивать векторы?

1-й способ вам уже известен, посчитать расстояние между векторами при помощи теоремы Пифагора

### Расстояние через т.Пифагора, евклидово расстояние

Как посчиать Евклидово расстояние между векторами `a(2, 2)` и `b(5, 6)`?

1. Надо посчитать расстояние по оси X:
X = $a_1 - b_1$ = 5 - 2 = 3

2. По оси Y:
Y = $a_2 - b_2$ = 6 - 2 = 4

3. Примененить теорему Пифагора:

$$
\sqrt{X^2 + Y^2} = \sqrt{3^2+4^2} = \sqrt{25} = 5
$$

Или в виде одной формулы:

$$
d = \sqrt{(a_1 - b_1)^2 + (a_2 - b_2)^2}
$$

````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import FancyBboxPatch

# Create a large figure for presentation
plt.figure(figsize=(14, 10))
plt.gca().set_aspect('equal', adjustable='box')

# Define the points
A = np.array([2, 2])
B = np.array([5, 6])
C = np.array([5, 2])  # Right angle point

# Calculate distances using Pythagorean theorem
horizontal_dist = abs(B[0] - A[0])  # |4-2| = 2
vertical_dist = abs(B[1] - A[1])    # |5-2| = 3
hypotenuse = np.sqrt(horizontal_dist**2 + vertical_dist**2)  # √(2² + 3²) = √13 ≈ 3.606

# Plot the triangle
plt.plot([A[0], C[0]], [A[1], C[1]], 'b-', linewidth=3, label=f'Horizontal side = {horizontal_dist}')
plt.plot([C[0], B[0]], [C[1], B[1]], 'g-', linewidth=3, label=f'Vertical side = {vertical_dist}')
plt.plot([A[0], B[0]], [A[1], B[1]], 'r-', linewidth=3, label=f'Hypotenuse = {hypotenuse:.3f}')

# Plot the points
plt.plot(A[0], A[1], 'ko', markersize=8, label=f'Point A ({A[0]}, {A[1]})')
plt.plot(B[0], B[1], 'ko', markersize=8, label=f'Point B ({B[0]}, {B[1]})')
plt.plot(C[0], C[1], 'ko', markersize=8, label=f'Point C ({C[0]}, {C[1]})')

# Mark the right angle
plt.plot(C[0], C[1], 'ro', markersize=10, markerfacecolor='none', markeredgewidth=2)

# Add distance labels with arrows
# Horizontal side
plt.annotate(f'{horizontal_dist}',
             xy=((A[0]+C[0])/2, (A[1]+C[1])/2),
             xytext=((A[0]+C[0])/2, (A[1]+C[1])/2 - 0.2),
             ha='center', fontsize=14, fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue", alpha=0.7))

# Vertical side
plt.annotate(f'{vertical_dist}',
             xy=((C[0]+B[0])/2, (C[1]+B[1])/2),
             xytext=((C[0]+B[0])/2 + 0.2, (C[1]+B[1])/2),
             ha='center', fontsize=14, fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.3", facecolor="lightgreen", alpha=0.7))

# Hypotenuse
plt.annotate(f'{hypotenuse:.3f}',
             xy=((A[0]+B[0])/2, (A[1]+B[1])/2),
             xytext=((A[0]+B[0])/2 - 0.3, (A[1]+B[1])/2 + 0.3),
             ha='center', fontsize=14, fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.3", facecolor="lightcoral", alpha=0.7))

# Add the Pythagorean theorem equation
equation_text = f'{horizontal_dist}² + {vertical_dist}² = {hypotenuse:.3f}²\n'
equation_text += f'{horizontal_dist**2} + {vertical_dist**2} = {hypotenuse**2:.3f}'
plt.annotate(equation_text,
             xy=(0.02, 0.98), xycoords='axes fraction',
             fontsize=16, fontweight='bold', va='top',
             bbox=dict(boxstyle="round,pad=0.5", facecolor="yellow", alpha=0.8))

# Set up the plot
plt.xlim(1.5, 5.5)
plt.ylim(1.5, 7)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='k', alpha=0.1)
plt.axvline(x=0, color='k', alpha=0.1)

# Add labels and title
plt.xlabel('X', fontsize=14, fontweight='bold')
plt.ylabel('Y', fontsize=14, fontweight='bold')
plt.title('Теорема Пифагора в 2d',
          fontsize=18, fontweight='bold', pad=20)

# Add legend
# plt.legend(loc='upper right', fontsize=12, framealpha=0.9)

# Make axes equal for proper right angle visualization


# Add grid for better measurement visualization
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig('pythagor.png')
plt.show()
````

Но что делать с 3-х мерными векторами, 10-мерными, 100-мерными? Формула остается такой же, только количество слагаемых меняется.

Пусть нам надо найти расстояние между векторами $a_n$ и $b_n$, где n-размерность вектора:
- $a = (a_1, a_2, \dots, a_n)$
- $b = (b_1, b_2, \dots, b_n)$.  

Тогда евклидово расстояние между ними вычисляется по формуле:

$$
d = \sqrt{(a_1 - b_1)^2 + (a_2 - b_2)^2 + \dots + (a_n - b_n)^2}
$$

Или, в компактной записи через знак суммы:

$$
d = \sqrt{\sum_{i=1}^{n} (a_i - b_i)^2}
$$

Чему равна диагональ n-мерного куба с длиной 1.

2: корень 2
3: корень из 3-х
400: 20
n: корень из n

### Косинусное расстояние

Второй способ чуть сложнее, но он лучше работает с векторами большой размерности. Посчитать косинусное расстояние

Косинуснове расстояние позволяет понять насколько 2 вектора смотрят в одно направление. Чем оно меньше, то более сонаправлены 2 вектора

Вычисляется оно так:

`1 - (cosinus угла между векторами)`

Пояснение:

````python
import plotly.graph_objects as go
from math import cos, sin, radians, sqrt, acos

# Fixed vector v1 at 45°
angle1 = 45
x1 = cos(radians(angle1))
y1 = sin(radians(angle1))

# Initial v2 aligned with v1 (angle2_init = 45°)
angle2_init = 45

# Rotation matrix function: rotate a vector (x,y) by delta radians
def rotate(x, y, delta):
    x_rot = x * cos(delta) - y * sin(delta)
    y_rot = x * sin(delta) + y * cos(delta)
    return x_rot, y_rot

# Compute quantities for a given absolute angle of v2
def compute_values(angle2):
    delta = radians(angle2 - angle1)      # rotation from v1 to v2
    x2, y2 = rotate(x1, y1, delta)        # rotate v1 to get v2
    dot = x1 * x2 + y1 * y2                # cos(theta) (both are unit)
    dot = max(-1.0, min(1.0, dot))         # clamp to avoid numerical issues
    theta_deg = acos(dot) * 180.0 / 3.141592653589793
    cos_d = 1 - dot
    return x2, y2, dot, d, cos_d, theta_deg

# Initial values
x2, y2, dot, d, cos_d, theta_deg = compute_values(angle2_init)

# Create figure
fig = go.Figure()

# Trace for v1 (fixed)
fig.add_trace(go.Scatter(
    x=[0, x1, None],
    y=[0, y1, None],
    mode='lines+markers',
    name='v1 (45°)',
    line=dict(color='blue', width=3),
    marker=dict(size=8, color='blue')
))

# Trace for v2 (will be updated)
fig.add_trace(go.Scatter(
    x=[0, x2, None],
    y=[0, y2, None],
    mode='lines+markers',
    name='v2',
    line=dict(color='red', width=3),
    marker=dict(size=8, color='red')
))

# Origin marker
fig.add_trace(go.Scatter(
    x=[0], y=[0],
    mode='markers',
    marker=dict(size=10, color='black'),
    name='origin',
    showlegend=False
))

# Initial annotation
annotation_text = (f"cos(θ) = {dot:.3f}<br>"
                   f"cos distance = {cos_d:.3f}<br>"
                   f"θ = {theta_deg:.1f}°")
fig.add_annotation(
    x=0.5, y=1.05, xref='paper', yref='paper',
    text=annotation_text,
    showarrow=False,
    font=dict(size=14, color='black'),
    align='center',
    bgcolor='lightgray',
    bordercolor='black',
    borderwidth=1
)

# Prepare slider steps
steps = []
angles = list(range(0, 361, 5))  # 0° to 360° in steps of 5°

for angle in angles:
    x2_new, y2_new, dot_new, d_new, cos_d_new, theta_new = compute_values(angle)
    new_annotation = (f"cos(θ) = {dot_new:.3f}<br>"
                      f"cos расстояние= {cos_d_new:.3f}<br>"
                      f"θ = {theta_new:.1f}°")

    # Data update: list of dicts for each trace (index 0: v1, 1: v2, 2: origin)
    data_update = {
        "x": [
                    [0, x1],  # Vector 1 (unchanged)
                    [0, x2_new],  # Vector 2 (updated)
                    None,       # Arc (updated)
                ],
                "y": [
                    [0, y1],  # Vector 1 (unchanged)
                    [0, y2_new],  # Vector 2 (updated)
                    None,       # Arc (updated)
                ],
    }


    # Layout update: update the annotation text
    # Copy original annotation properties and change 'text'
    orig_ann = fig.layout.annotations[0]
    new_ann = {
        'x': orig_ann.x,
        'y': orig_ann.y,
        'xref': orig_ann.xref,
        'yref': orig_ann.yref,
        'text': new_annotation,
        'showarrow': orig_ann.showarrow,
        'font': orig_ann.font,
        'align': orig_ann.align,
        'bgcolor': orig_ann.bgcolor,
        'bordercolor': orig_ann.bordercolor,
        'borderwidth': orig_ann.borderwidth
    }
    layout_update = {'annotations': [new_ann]}

    step = dict(
        method='update',
        args=[data_update, layout_update],
        label=f'{angle}°'
    )
    steps.append(step)

# Create slider
sliders = [dict(
    active=angles.index(angle2_init),
    currentvalue=dict(prefix=''),
    pad=dict(t=50),
    steps=steps
)]

# Update layout
fig.update_layout(
    title='Демонстрация косинусного расстояния',
    xaxis=dict(range=[-1.2, 1.2], zeroline=True, zerolinecolor='black'),
    yaxis=dict(range=[-1.2, 1.2], zeroline=True, zerolinecolor='black'),
    width=700,
    height=700,
    showlegend=True,
    sliders=sliders,
    hovermode='closest',
    xaxis_title='x',
    yaxis_title='y',
    yaxis_scaleanchor='x',  # equal aspect ratio
    plot_bgcolor='white',
    margin=dict(l=50, r=50, t=100, b=50)
)

# Add grid lines
fig.update_xaxes(showgrid=True, gridwidth=1, gridcolor='lightgray')
fig.update_yaxes(showgrid=True, gridwidth=1, gridcolor='lightgray')

# Show figure
fig.show()
````

Но как вычислить косинус между 2-мя векторами?

$$
|a| * |b| * \cos{a, b} = a_1b_1 + a_2b_2
$$

$$
\cos{a, b} = \frac{a_1b_1 + a_2b_2}{ |a| * |b|}
$$

#### Формула для 2d

Для начала рассмотрим формулу для 2-мерной плоскости

Формула косинусного расстояния (косинуса угла) для двумерных векторов выглядит как частный случай общей формулы. Для векторов $\mathbf{a} = (a_1, a_2)$ и $\mathbf{b} = (b_1, b_2)$:

$$
\cos(\mathbf{a}, \mathbf{b}) = \frac{a_1 b_1 + a_2 b_2}{\sqrt{a_1^2 + a_2^2} \cdot \sqrt{b_1^2 + b_2^2}}
$$

где:
- $a_1 b_1 + a_2 b_2$ — скалярное произведение векторов,
- $\sqrt{a_1^2 + a_2^2}$ — длина (модуль) вектора $\mathbf{a}$,
- $\sqrt{b_1^2 + b_2^2}$ — длина вектора $\mathbf{b}$.


Рассмотрим векторы $\mathbf{a} = (3, 4)$ и $\mathbf{b} = (4, 3)$.

1. **Скалярное произведение**:  
   $3 \cdot 4 + 4 \cdot 3 = 12 + 12 = 24$

2. **Длина вектора $\mathbf{a}$**:  
   $\sqrt{3^2 + 4^2} = \sqrt{9 + 16} = \sqrt{25} = 5$

3. **Длина вектора $\mathbf{b}$**:  
   $\sqrt{4^2 + 3^2} = \sqrt{16 + 9} = \sqrt{25} = 5$

4. **Косинус угла**:  
   $\cos(\mathbf{a}, \mathbf{b}) = \frac{24}{5 \cdot 5} = \frac{24}{25} = 0.96$

Таким образом, косинус угла между векторами $(3,4)$ и $(4,3)$ равен $0.96$, что соответствует острому углу примерно $16.26^\circ$.

#### Формула для размерности n

$$
\cos(\mathbf{a}, \mathbf{b}) = \frac{a_1b_1 + a_2b_2 + \dots + a_nb_n}{\sqrt{a_1^2+a_2^2+\dots+a_n^2} *  \sqrt{b_1^2+b_2^2+\dots+b_n^2}}
$$

где:

- $a_1b_1 + a_2b_2 + \dots + a_nb_n$ - скалярное произдедение векторов
- $\sqrt{a_1^2+a_2^2+\dots+a_n^2}$ - длина вектора a
- $\sqrt{b_1^2+b_2^2+\dots+b_n^2}$ - длина вектора b

````python
def is_zero(vector):
    for i in vector:
        if i != 0:
            return False
    return True

is_zero()
not(is_zero())
````
