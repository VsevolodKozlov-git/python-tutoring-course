# Урок 27. Получаем слова и приводим их к единой форме

### Зачем приводить слова к единой форме?



Наш поисковик будет искать совпадение слов в поисковом запросе и тексте. Чем больше слов совпадают, тем более подходящий результат.

Чтобы наш поисковик работал правильно, необходимо, чтобы слова в запросе и текстах были одной форме. Приведу пример.


Запрос: `Песня, где друзья идут домой, но встречают разбойников`


Текст:
```
Два друга шли домой
Дорогой ночной.
Вдруг разбойники из леса
Вышли целою толпой.
```

Если сравнивать без приведения к нормальной форме, тут совпадает только слово `домой`, остальные слова не совпадают.


Если привести к нормальной форме, то найдутся:
- дом
- друг
- идти
- разбойник


### Получить данные с прошлого урока

````python
import json

with open('tracks_full.json') as file:
    texts_base = json.load(file)
````

### Получить список слов

````python
text = texts_base[0]['text']
print(text)
````

````python
text = 'кое-как - привет -  - '
text = ''.join([i for i in text if i.isalpha() or ord(i) in (32, 160, 45) or i == '\n'])
text_splited = text.split()
text_splited = [i for i in text_splited if i != '-']
text_splited
````

````python
import re
re.findall(r'\w[\w\-]*', 'кое-как - привет -  - ')
````

````python
def get_words(text):
    return re.findall(r'\w[\w\-]*', text)
````

````python
get_words(texts_base[0]['text'])
````

### Привести слова к нормальной форме

````python
!pip install pymorphy3
````

````python
from pymorphy3 import MorphAnalyzer

analyzer = MorphAnalyzer()

parsed = analyzer.parse('горочки')

parsed[0].normal_form
````

````python
def get_normal_form(word):
    return analyzer.parse(word.lower())[0].normal_form

words = get_words(texts_base[0]['text'])
words_normalized = [get_normal_form(word) for word in words]
for word_base, words_normalized in zip(words, words_normalized):
    print(f'{word_base}: {words_normalized}')
````

### Проанализировать слова

````python
texts_normalized = []
for entry in texts_base:
    text = entry['text']
    title = entry['title']
    words = get_words(text)
    words_normalized = [get_normal_form(word) for word in words]
    texts_normalized.append({'title': title, 'words': words_normalized})

````

````python
from collections import Counter

words_counter = Counter()
for entry in texts_normalized:
    words_counter.update(entry['words'])

words_counter.most_common(100)
````

### Убрать лишние слова

````python
with open('russian') as file:
    stop_words = file.read().split()

stop_words = stop_words + ['припев', 'человек', 'парень']
stop_words = [get_normal_form(word) for word in  stop_words]
stop_words
````

````python
texts_without_stop = []
for entry in texts_normalized:
    words = entry['words']
    title = entry['title']
    words_without_stop = [word for word in words if word not in stop_words]
    texts_without_stop.append({'title': title, 'words': words_without_stop})

texts_without_stop[0]
````

### Сохранить результат

````python
with open('words_clear.json', 'w', encoding='utf-8') as file:
    json.dump(texts_without_stop, file, ensure_ascii=False, indent=2)
````
